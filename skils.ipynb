## 문제 풀이 초기에 BigO 생각하기
### 문제풀이 순서
* 문제의 데이터 개수를 본다.
* 실행시간을 확인(보통 1초 = $2 * 10^7$ 회 (2000만회)
* 주어진 시간 안에서 사용할 수 있는 알고리즘을 생각
* 해당 알고리즘으로 풀이 방법을 구상해본다

### 1.문제 이해하는법
* 예시를 통해, 규칙을 이해
* 준제에 적힌 예시, 실행 예시를 모두 손/머리로 풀 수 있어야함
* 그래도 이해 안된다면 예시를 만든다
* 예시 만드는 법
  * 데이터 개수를 1개 혹은 아주 작게 만듦
  * 적용되는 규칙을 줄여본다
* 예시를 풀어보면서 일반화 규칙을 찾고, 그 일반화 규칙을 코드로 작성

### 2. 모듈화
* 전체 큰 문제를, 내가 다룰 수 잇는 모듈로 쪼개주는 것
* 모듈 만드는 법
  * input / output 결정
  * 연산결정

### 3. 디버깅
* 로그찍을때는 이진탐색느낌으로
  * 중간 부분에 먼저 찍고 로직 잘 흐르는지 확인
  * 의심 지점 로그찍어서 확인

### 재귀적 문제해결 방식을 이해해야만 활용할 수 있는 알고리즘
* 알고리즘 개념들 (이진탐색, 분할 정렬, DFS, merge sort등)

### Big-O
#### 암기
* 코테에서 대개 1초에 2000만개의 실행횟수를 처리한다

#### O(1)
* 반복문 없이 단순연산, 배열 인덱스 접근, 해시 맵 조회등
#### O(logN)
* 반복마다 처리해야하는 데이터의 수가 절반씩 줄어드는 경우
#### O(N)
* 데이터를 처음부터 끝까지 한 번 훑는 반복문
#### O($N^2$)
* 이차 반복문

#### 성능 순서 암기(빠름 -> 느림)
* O(1) -> O($logN$) -> O($N$) -> O($NlogN$) -> O($N^2$) -> O($2^n$) -> O($n!$)
* na ->  $10^7$  ->     백만개 ->      10만개 ->    2000 ~ 5000 ->  20 ~ 25  -> 10 ~ 11



### 각 Big-O별 대표 알고리즘
* O(1): 해시
* O($logN$): 이진탐색
* O(N): 선형탐색
* O($NlogN$): 정렬
* O($N^2$): 브루트포스
