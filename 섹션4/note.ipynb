## 해시
### 해시 함수
* 임의의 길이를 가지는 임의의 데이터를 입력받아서, 고정된 길이의 해시 코드로 변환하는 함수

### 해시 테이블
* 저장하려는 데이터의 해시 값을 인덱스처럼 사용
* 효율적인 데이터 검색, 삽입, 삭제 가능 - 데이터 크기와 상관없이 $O(1)$

### 해시 충돌
* 서로 다른 입력에 대해 동일한 해시 코드가 생성될 수 있음
* 파이썬 set, dictionary -> 내부적으로 해시 충돌에 대한 대책 있음

### 해시 가능한 자료형
#### 불변 이거나, 내용이 변경되지 않는 자료형
#### 해시 가능한 자료형은 __hash__() 메서드 가짐
* 숫자형
* 문자열
* 튜플
* Boolean
* None

### 해시 불가능한 자료형
#### 가변이거나, 내부의 값이 변경될 수 있는 자료형
* 리스트(list)
* 셋(Set)
* 딕셔너리(Dictionary)


```python
# 예시: 리스트를 셋으로 변환
my_list = ['a', 'b', 'c', 'a']
my_set = list(my_list)
print(my_list)
print(my_set)
```
